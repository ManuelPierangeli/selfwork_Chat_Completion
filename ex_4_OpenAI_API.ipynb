**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile.

```python
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
import json

nome = input("Inserisci il nome: ")
eta = input("Inserisci l'età: ")
citta = input("Inserisci la città: ")
professione = input("Inserisci la professione: ")

#dizionario
data = {
    "name": nome,
    "age": eta,
    "city": citta,
    "profession": professione
}

data_str = json.dumps(data, ensure_ascii=False)

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente esperto nella generazione di contenuti testuali naturali. "
                "Riceverai in input un set di dati strutturati (in formato JSON o dizionario). "
                "Il tuo compito è trasformarli in una descrizione testuale fluida e naturale. "
                "Regole da seguire: "
                "1. Includi tutte le informazioni chiave presenti nei dati. "
                "2. Mantieni un tono chiaro, professionale e scorrevole. "
                "3. Non aggiungere informazioni non presenti. "
                "4. Restituisci esclusivamente il testo descrittivo finale, senza frasi introduttive."
            )
        },
        {
            "role": "user",
            "content": data_str
        }
    ],
)

print(completion.choices[0].message.content)
```
