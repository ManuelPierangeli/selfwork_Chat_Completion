**Traccia:** Scrivi un prompt che generi domande di comprensione su un breve testo.

**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente didattico ed esperto nella creazione di test di verifica."
                "Riceverai in input un breve testo. Il tuo compito è generare un insieme di domande di comprensione basate esclusivamente sul contenuto fornito."
                "Regole da seguire: "
                "1. Crea 3 domande a scelta multipla (con opzioni A, B, C, D) oppure a risposta aperta basate sul testo."
                "2. Includi anche le relative soluzioni o risposte corrette alla fine del test."
                "3. Mantieni le domande e le risposte concise e pertinenti al testo."
                "4. Non aggiungere domande o informazioni che non siano presenti nel testo fornito."
                "5. Mantieni un tono chiaro e didattico, adatto ad un contesto educativo."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(complention.choices[0].message.content)
```
