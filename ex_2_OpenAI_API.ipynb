**Traccia:** Crea un prompt che estragga i dettagli chiave (nome, età, professione) da un testo descrittivo su una persona.

**Obiettivo:** Estrarre informazioni strutturate da un testo libero.

```python

import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI(
    api_key=os.getenv('OPENAI_API_KEY'),
    base_url='https://generativelanguage.googleapis.com/v1beta/openai/',
    )
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gemini-3.5-flash",
    messages=[
        {
            "role" : "system",
            "content" : ""
            "Sei un assistente esperto nell'estrazione di informazioni da testi.""Leggi attentamente il testo fornito ed estrai esclusivamente i seguenti dettagli chiave: "
            "Nome e Cognome, Età, Professione. "
            "Restituisci l'output in modo pulito e ordinato. "
            "Se un dettaglio non è presente, scrivi 'Non specificato'."
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(complention.choices[0].message.content)
```
