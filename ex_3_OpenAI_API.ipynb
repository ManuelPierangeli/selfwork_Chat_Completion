**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI(
    api_key=os.getenv('OPENAI_API_KEY'),
    base_url='https://generativelanguage.googleapis.com/v1beta/openai/',
    )
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gemini-3.5-flash",
    messages=[
        
        {
            "role" : "system",
            "content" : "Sei un giornalista esperto nella sinstesi di notizie."
            "Riassumi l'articolo fornito in unico paragrafo chiaro e scorrevole"
            "Regole da rispettare:"
            "1. Non superare i 255 caratteri (spazi inclusi)."
            "2. Usa un solo paragrafo (niente elenchi puntati,titoli)."
            "3. Cattura solo i fatti principali"
            "4. Restituisci esclusivamente il testo del riassunto."
        },
        {
            "role" : "user",
            "content" : input_text
        }

    ],
)

print(complention.choices[0].message.content)
```
